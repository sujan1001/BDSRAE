# Multilingual Retrieval Fairness - results notebook

This notebook reads the reproducible smoke outputs. It intentionally labels them as synthetic smoke evidence, not final benchmark results. Run the pipeline first from the repository root.

```python
import csv
import json
from pathlib import Path

ROOT = Path.cwd()
RESULTS = ROOT / 'results' / 'smoke'
metrics = list(csv.DictReader((RESULTS / 'metrics.csv').open(encoding='utf-8')))
fairness = json.loads((RESULTS / 'fairness_summary.json').read_text(encoding='utf-8'))
systems = json.loads((RESULTS / 'system_metrics.json').read_text(encoding='utf-8'))
manifest = json.loads((RESULTS / 'run_manifest.json').read_text(encoding='utf-8'))
print(manifest['run_scope'], manifest['data_validation'])
```

```python
experiment_ids = sorted({row['experiment_id'] for row in metrics})
for experiment_id in experiment_ids:
    print('\n', experiment_id)
    summary = fairness[experiment_id]
    for language in summary['languages']:
        value = summary['language_metrics'][language]['recall@10']['mean']
        print(language, 'Recall@10=', round(value, 3))
    print('gap Recall@10=', round(summary['gap_max_minus_min']['recall@10'], 3))
```

```python
print('Systems scale sweep')
for row in systems['scale_sweep']:
    print(row)
print('Spark probe:', systems['spark'])
```
